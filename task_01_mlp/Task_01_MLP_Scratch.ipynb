# Task 1: Vectorized Multi-Layer Perceptron (MLP) from Scratch
Implementing batch-wise forward pass, softmax cross-entropy loss, and matrix calculus backprop using NumPy.


```python
import numpy as np
from mlp_scratch import MLP
np.random.seed(42)
print('NumPy environment ready.')
```

```python
# Generate synthetic dataset
X = np.random.randn(256, 10)
y = np.random.randint(0, 3, size=(256,))
print(f'Input X: {X.shape}, Target y: {y.shape}')
```

```python
# Build and train 3-layer MLP
model = MLP(layer_sizes=[10, 32, 16, 3])
for epoch in range(1, 101):
    logits = model.forward(X)
    loss = model.criterion.forward(logits, y)
    model.backward(model.criterion.backward())
    model.update_params(0.1)
    if epoch % 20 == 0:
        acc = np.mean(np.argmax(logits, axis=1) == y)
        print(f'Epoch {epoch:03d} | Loss: {loss:.4f} | Accuracy: {acc*100:.2f}%')
```
